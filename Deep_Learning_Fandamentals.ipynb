#**What is Neuron ?**

A neuron is a unit in a neural network that **`computes a weighted sum of inputs, adds a bias, and applies an activation function`** to produce an output, simulating decision-making processes.

#**What is weight ?**

**A weight is a `trainable parameter` that determines the `strength and importance` of a `connection between neurons`.**

#**What is Hidden Layer ?**

**A hidden layer is an `intermediate layer of neurons` that `receives inputs from the previous layer`, applies weights, biases, and activation functions, and `passes the transformed outputs to the next layer`, enabling `feature extraction and pattern learning.`**

#**What is Forward Propagation ?**

**Forward propagation is the `sequential process` where `input data is passed through each layer of a neural network,` with neurons computing weighted sums of inputs, adding biases, applying activation functions, and producing outputs that flow to the next layer until the final prediction is obtained.**

#**What is Back Propagation ?**

**Backpropagation is the process of computing the `gradient of the loss function` with respect to the network's weights and biases by `propagating the error backward through the layers.` It is used to `update the weights via optimization algorithms` like gradient descent to `minimize the loss`.**

#**What is Memoization ?**

**Memoization is an `optimization technique` where `intermediate computational results are cached and reused to avoid redundant calculations,` often applied in dynamic programming or recursive algorithms rather than directly in neural network operations.**

#**What is Activation Function ?**

An activation function is a **`mathematical function applied to the weighted sum of inputs and bias in a neuron to introduce non-linearity`**, ensuring the neural network can learn and represent complex patterns beyond linear relationships.

#**What is Dead Activation ?**

Dead activation occurs **`when a neuron consistently outputs zero due to the activation function`**, making it unresponsive to input variations. This commonly happens with **`the ReLU activation function when the weighted sum of inputs is negative, causing gradients to be zero, preventing weight updates, and effectively deactivating the neuron during training.`**

#**What is Exploding Gradieant Problem ?**

The exploding gradient problem occurs **`when gradients grow exponentially large during backpropagation,`** causing extreme weight updates and instability in training. This typically happens in deep networks or recurrent neural networks (RNNs) when weight updates compound across many layers, **`leading to divergence instead of convergence.`**

**Causes:**

&#10687; Large weight initialization.

&#10687; High learning rates.

&#10687; Unstable recursive gradient multiplications in deep architectures.

**Solutions:**

&#10687; `Gradient clipping:` Restricts gradients to a fixed threshold.

&#10687; `Weight regularization:` Uses techniques like L2 regularization.

&#10687; `Proper weight initialization:` Methods like Xavier or He initialization help stabilize gradients.

&#10687; `Smaller learning rates:` Helps control large updates.

#**What is DropOut ?**

Dropout is a **`regularization technique`** that randomly deactivates (drops) a fraction of neurons in a layer during training, **`preventing overfitting`**. At each training step, **`neurons are set to zero`** with a probability p, while the remaining neurons are scaled up to maintain expected outputs.

**Key Properties:**

&#10687; Helps prevent overfitting by forcing the network to learn robust features.

&#10687; Improves generalization by reducing reliance on specific neurons.

&#10687; Commonly used in fully connected layers of deep networks.

#**What is Early Stoping ?**

Early Stopping is a **`regularization technique`** that halts training when the model's performance on a **`validation set starts to degrade, preventing overfitting`**. It monitors a chosen metric (e.g., **`validation loss`**) and **`stops training if no improvement`** is observed for a predefined number of epochs (patience).

**Key Properties:**

&#10687; Prevents overfitting by **`stopping before the model memorizes`** training data.

&#10687; **`Saves computational resources`** by avoiding unnecessary epochs.

&#10687; Requires a validation set to track generalization performance.

**Implementation:**

&#10687; **`Monitor:`** Track validation loss (or accuracy).

&#10687; **`Patience:`** **`Allow a few epochs`** without improvement before stopping.

&#10687; **`Restore Best Model:`** Optionally revert to the model with the lowest validation loss.

**Common Use:**

Often combined with learning rate scheduling to **`ensure the model converges before stopping.`**

#**What is loss Function ?**

A loss function **`measures the error between the model’s predictions and actual targets,`** guiding optimization. Common types include MSE for regression and Cross-Entropy for classification.

#**What is Cost Function ?**

A cost function **`calculates the average loss across all training samples`**, guiding weight updates. It helps **`optimize the model by minimizing overall error`** using methods like **`Gradient Descent`**.

**Key Differences:**

`Loss Function:` Measures error for a single sample.

`Cost Function:` Computes the average loss over all training samples.
